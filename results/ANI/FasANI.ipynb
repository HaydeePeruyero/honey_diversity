```python
import pandas as pd

# Define la ruta al archivo
ruta_archivo = "fastani.out.corredted"

# Lee el archivo como un DataFrame
df = pd.read_csv(ruta_archivo, sep="\t", header=None)

# Asigna nombres a las columnas
df.columns = ["query", "reference", "ani", "cbfm", "tqf"]
```

```python
df
```

```python
resultado = df[(df['query'] == '46_Melli.022') & (df['reference'] == '46_Melli.022')]

# Mostrar el resultado
print(resultado)
```

```python
# Obtén los IDs únicos de query y reference
ids_unicos = pd.unique(df[['query', 'reference']].values.ravel())

# Crea un DataFrame vacío con los IDs únicos como filas y columnas, inicializado en 0
matriz_ani = pd.DataFrame(0, index=ids_unicos, columns=ids_unicos)

# Llena la matriz con los valores de `ani`
for _, row in df.iterrows():
    matriz_ani.at[row['query'], row['reference']] = row['ani']
```

```python
# Muestra la matriz
matriz_ani
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Configuración del tamaño del gráfico
plt.figure(figsize=(20, 16))

# Crear el heatmap con seaborn
sns.heatmap(matriz_ani, annot=True, fmt=".2f", cmap="viridis", cbar=True)

# Títulos y etiquetas
plt.title("Heatmap de ANI", fontsize=16)
plt.xlabel("Reference", fontsize=12)
plt.ylabel("Query", fontsize=12)

# Mostrar el heatmap
plt.tight_layout()
# Guardar el gráfico como SVG
plt.savefig("heatmap_ani.svg", format="svg")
plt.show()
```

```python
groups = {
    "Group_1": [0, 1,2,3],
    "Group_2": [4,5,14],
    "Group_3": [10,11,12,13],
    "Group_4": [6,7,8,9]
}
```

```python
import numpy as np
import pandas as pd
from itertools import combinations
# Calcular media y varianza dentro de cada grupo
results_within = {}
for group_name, indices in groups.items():
    submatrix = matriz_ani.iloc[indices, indices]
    results_within[group_name] = {
        "mean": submatrix.values.mean(),
        "std_dev": submatrix.values.std(),
        "var": submatrix.values.var()
    }

# Calcular media y varianza entre grupos
results_between = {}
for (group_name_1, indices_1), (group_name_2, indices_2) in combinations(groups.items(), 2):
    submatrix = matriz_ani.iloc[indices_1, indices_2]
    key = f"{group_name_1} vs {group_name_2}"
    results_between[key] = {
        "mean": submatrix.values.mean(),
        "std_dev": submatrix.values.std(),
        "var": submatrix.values.var()
    }

# Mostrar resultados
print("Media y Varianza Dentro de Grupos:")
print(pd.DataFrame(results_within).T)

print("\nMedia y Varianza Entre Grupos:")
print(pd.DataFrame(results_between).T)
```

```python
# Crear el clustermap
sns.clustermap(
    matriz_ani,
    annot=True,         # Muestra los valores en las celdas
    fmt=".2f",          # Formato de los valores (dos decimales)
    cmap="viridis",     # Paleta de colores
    cbar=True,          # Incluye la barra de color
    metric="euclidean", # Métrica para calcular las distancias
    method="average",   # Método de enlace para clustering
    figsize=(20, 16)     # Tamaño del gráfico
)

# Mostrar el clustermap
plt.title("Clustermap de ANI")
plt.show()
```
